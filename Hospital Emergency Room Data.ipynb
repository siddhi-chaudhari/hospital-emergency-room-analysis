```python
import pandas as pd

df = pd.read_csv('Hospital Emergency Room Data.csv')
```

```python
df.head()
```

```python
df['Department Referral']=df['Department Referral'].fillna(df['Department Referral'].mode()[0])
```

```python
df['Patient Satisfaction Score']=df['Patient Satisfaction Score'].fillna(df['Patient Satisfaction Score'].median())
```

```python
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df=df.rename(columns={'patient_admission_date':'admission_date'})
df=df.rename(columns={'first_inital':'first_name'})
```

```python
df.columns
```

```python
df.rename(columns={'id': 'patient_id'}, inplace=True)

print(df.columns.tolist())
```

```python
labels=['Young Adult','Adult','Middle-Aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python

# Remove the duplicate admission flag column
df.drop(columns=['admission_flag.1'], inplace=True)

# Check the remaining columns
print(df.columns.tolist())

```

```python
df.head()
```

```python

# Convert admission_date into datetime format
df['admission_date'] = pd.to_datetime(
    df['admission_date'],
    dayfirst=True,
    errors='coerce'
)

# Create separate date and time columns
df['admission_time'] = df['admission_date'].dt.strftime('%H:%M')
df['admission_date'] = df['admission_date'].dt.strftime('%d-%m-%Y')

# Display the result
display(df[['patient_id', 'admission_date', 'admission_time']].head(10))

```

```python

# Correct patient IDs with sequential numbers
df['patient_id'] = range(1, len(df) + 1)

# Display the first 10 patient IDs
display(df[['patient_id']].head(10))

# Check that all IDs are unique
print("Total patient records:", len(df))
print("Unique patient IDs:", df['patient_id'].nunique())
```

```python
df.head()
```

```python

# Merge date and time into one column
df['admission_date'] = pd.to_datetime(
    df['admission_date'].astype(str) + ' ' +
    df['admission_time'].astype(str),
    format='%d-%m-%Y %H:%M',
    errors='coerce'
)

# Remove the separate time column
df.drop(columns=['admission_time'], inplace=True)

# Display the result
display(df[['patient_id', 'admission_date']].head(10))
```

```python

# Standardize gender values
df['gender'] = (
    df['gender']
    .astype('string')
    .str.strip()
    .str.lower()
    .replace({
        'm': 'Male',
        'male': 'Male',
        'f': 'Female',
        'female': 'Female'
    })
)

# Check the results
print(df['gender'].value_counts(dropna=False))
```

```python

# Rename waiting time column to indicate minutes
df.rename(
    columns={'waittime': 'waiting_time_minutes'},
    inplace=True
)

# Display the result
display(df[['patient_id', 'waiting_time_minutes']].head(10))
```

```python

# Display waiting time with the 'min' suffix
df['waiting_time_minutes'] = (
    df['waiting_time_minutes'].astype(str) + ' min'
)

# Display the result
display(df[['patient_id', 'waiting_time_minutes']].head(10))
```

```python

df.rename(
    columns={'waiting_time_minutes': 'waiting_time'},
    inplace=True
)

display(df[['patient_id', 'waiting_time']].head(10))
```
